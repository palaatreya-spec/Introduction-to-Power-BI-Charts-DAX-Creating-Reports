**Question 1 : Define Power BI and What are the key components of the Power BI ecosystem? Briefly
explain:
● Power BI Desktop
● Power BI Service
● Power BI Mobile
● Power BI Gateway**

 - Answer: Power BI is a business intelligence and data visualization tool developed by Microsoft that helps users collect, transform, analyze, and visualize data to make data-driven decisions. It allows users to create interactive reports and dashboards from multiple data sources in a simple and efficient way.
 Power BI Desktop is a Windows-based application used to connect to data, clean and transform it using Power Query, create data models, write DAX measures, and design reports. Power BI Service is a cloud-based platform used to publish, share, collaborate, and schedule data refreshes for reports and dashboards. Power BI Mobile allows users to access dashboards and reports on smartphones and tablets, enabling real-time insights on the go. Power BI Gateway acts as a bridge between on-premises data sources and Power BI Service, ensuring secure data refresh without moving data to the cloud.

**Question 2 : Compare the following Power BI visuals:
● Pie Chart vs Donut Chart
● Bar Chart vs Column Chart
When would you prefer one over the other? Give one example for each pair.**

 - Answer: A Pie Chart and Donut Chart both display part-to-whole relationships. A **Pie Chart** is simple and best used when there are very few categories, while a **Donut Chart** provides better readability by leaving space in the center for labels or KPIs. For example, a Donut Chart is preferred when showing sales percentage by region with a total sales value in the center.
 A **Bar Chart** displays data horizontally and is preferred when category names are long or when comparing many categories. A **Column Chart** displays data vertically and is best for showing trends over time or comparing fewer categories. For example, a Bar Chart is ideal for comparing sales by sub-category, while a Column Chart works well for monthly sales trends.

**Question 3 : Explain the significance of:
● Star schema vs Snowflake schema
● Primary key vs Foreign key in relationships (Power BI)
Why is cardinality important?**

 - Answer: A **Star Schema** is a data modeling structure where a central fact table is connected to multiple dimension tables, offering better performance and simpler relationships. A **Snowflake Schema** is a more complex structure where dimension tables are further normalized, which can increase complexity and reduce performance in Power BI.
 A **Primary Key** uniquely identifies records in a table, while a **Foreign Key** creates a relationship by linking one table to another. **Cardinality** defines the nature of relationships between tables (one-to-one, one-to-many, many-to-many). Correct cardinality is essential in Power BI because it ensures accurate data filtering, aggregation, and prevents incorrect calculations in reports.

**Question 4 : Differentiate between:
● Calculated column vs Measure
Also, define Row context and Filter context with simple examples.**

 - Answer: A **Calculated Column** is evaluated row by row during data refresh and stores values in the table, making it suitable for static calculations such as profit per row. A **Measure** is calculated dynamically based on filter context and is used for aggregations like total sales or average profit.
 **Row Context** refers to calculations performed on each row individually, such as calculating profit for every transaction. **Filter Context** refers to filters applied by visuals, slicers, or report interactions, such as calculating total sales for a selected region or time period.

**Question 5: What is the difference between a report and a dashboard in Power BI?**

 - Answer: A **Power BI Report** is a multi-page, interactive document created in Power BI Desktop that allows deep data exploration using filters, slicers, and drill-through features. A **Dashboard** is a single-page view created in Power BI Service that highlights key metrics by pinning visuals from one or more reports. Reports are used for detailed analysis, while dashboards are used for quick monitoring and high-level insights.

**Question 8 : Implement a DAX measure that calculates the percentage of total sales by product category.**

 - Answer: Sales % by Category =
DIVIDE(
    SUM('Global_superstore2'[Sales]),
    CALCULATE(
        SUM('Global_superstore2'[Sales]),
        ALL('Global_superstore2'[Category])
    )
)
